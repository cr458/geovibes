```python
import pandas as pd
import numpy as np
import geopandas as gpd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import glob
import os
from pathlib import Path
import matplotlib.pyplot as plt
import datashader as ds
import datashader.transfer_functions as tf
import datashader.colors as dc
import colorcet as cc
import holoviews as hv
from holoviews.operation.datashader import datashade, dynspread
hv.extension('bokeh')
# Load all parquet files from the Alabama embeddings directory
data_dir = "/home/christopherren/geovibes/embeddings/alabama_google"
parquet_files = glob.glob(os.path.join(data_dir, "*.parquet"))

print(f"Found {len(parquet_files)} parquet files")
print("Files:", [os.path.basename(f) for f in parquet_files])
# Read and examine structure of first file to understand data format
sample_df = pd.read_parquet(parquet_files[0])
print("Columns:", sample_df.columns.tolist())
print("Shape:", sample_df.shape)
print("Data types:")
print(sample_df.dtypes)
print("\nFirst few rows:")
print(sample_df.head())

```

```python
# Load all parquet files and combine them
all_dataframes = []

for i, file_path in enumerate(parquet_files):
    print(f"Loading file {i+1}/{len(parquet_files)}: {os.path.basename(file_path)}")
    df = gpd.read_parquet(file_path)
    all_dataframes.append(df)

# Combine all dataframes
combined_df = gpd.GeoDataFrame(pd.concat(all_dataframes, ignore_index=True))
print(f"\nCombined dataset shape: {combined_df.shape}")
print(f"Total number of points: {len(combined_df):,}")
```

```python
# Extract embeddings and prepare for PCA
# Assuming the embedding column contains arrays/lists
embeddings_array = np.stack(combined_df['embedding'].values)
print(f"Embeddings array shape: {embeddings_array.shape}")

# Standardize the embeddings
scaler = StandardScaler()
embeddings_scaled = scaler.fit_transform(embeddings_array)

print(f"Scaled embeddings shape: {embeddings_scaled.shape}")
print(f"Embeddings mean: {embeddings_scaled.mean():.6f}")
print(f"Embeddings std: {embeddings_scaled.std():.6f}")
```

```python
# Perform PCA to get the first 9 components
print("Performing PCA...")
pca = PCA(n_components=9)
pca_result = pca.fit_transform(embeddings_scaled)

print(f"PCA result shape: {pca_result.shape}")
print(f"Explained variance ratio for first 9 components:")
for i, var_ratio in enumerate(pca.explained_variance_ratio_):
    print(f"  PC{i+1}: {var_ratio:.4f} ({var_ratio*100:.2f}%)")

print(f"\nCumulative explained variance: {pca.explained_variance_ratio_.cumsum()[-1]*100:.2f}%")
```

```python
# Extract lat/lon coordinates from geometry column
# Convert to GeoDataFrame if not already
if not isinstance(combined_df, gpd.GeoDataFrame):
    combined_gdf = gpd.GeoDataFrame(combined_df)
else:
    combined_gdf = combined_df.copy()

# Extract longitude and latitude
combined_gdf['longitude'] = combined_gdf.geometry.x
combined_gdf['latitude'] = combined_gdf.geometry.y

print(f"Longitude range: {combined_gdf['longitude'].min():.6f} to {combined_gdf['longitude'].max():.6f}")
print(f"Latitude range: {combined_gdf['latitude'].min():.6f} to {combined_gdf['latitude'].max():.6f}")
```

```python
# Add PCA components to the dataframe
pc_columns = [f'PC{i+1}' for i in range(9)]
for i, pc_col in enumerate(pc_columns):
    combined_gdf[pc_col] = pca_result[:, i]

print("Added PCA components to dataframe")
print(f"New dataframe shape: {combined_gdf.shape}")
print(f"PCA columns: {pc_columns}")
```

```python
# Create summary statistics for the PCA components
pc_stats = pd.DataFrame({
    'Component': pc_columns,
    'Explained_Variance_Ratio': pca.explained_variance_ratio_,
    'Explained_Variance_Ratio_Pct': pca.explained_variance_ratio_ * 100,
    'Cumulative_Variance_Pct': np.cumsum(pca.explained_variance_ratio_) * 100,
    'Mean': [combined_gdf[col].mean() for col in pc_columns],
    'Std': [combined_gdf[col].std() for col in pc_columns],
    'Min': [combined_gdf[col].min() for col in pc_columns],
    'Max': [combined_gdf[col].max() for col in pc_columns]
})

print("Principal Component Analysis Summary:")
print("=" * 80)
print(pc_stats.round(4))
```

```python
import matplotlib.pyplot as plt
from concurrent.futures import ThreadPoolExecutor
import threading

# Sub-sample every 10th data point
sampled_gdf = combined_gdf.iloc[::10].copy()
print(f"Sub-sampled data: {len(sampled_gdf)} points from {len(combined_gdf)} total")

# Create plots for all 9 PC components using geopandas
fig, axes = plt.subplots(3, 3, figsize=(12, 15))
axes = axes.flatten()

# Define colormaps to cycle through
colormaps = ['coolwarm', 'RdYlBu', 'Spectral', 'RdBu', 'bwr', 'seismic', 'RdGy', 'PiYG', 'PRGn']

# Thread-local lock for matplotlib
plot_lock = threading.Lock()

def create_plot(i, pc_col):
    print(f"Creating geopandas plot for {pc_col}...")
    variance_pct = pca.explained_variance_ratio_[i] * 100
    title = f'{pc_col} (Explains {variance_pct:.2f}% of variance)'
    
    # Cycle through colormaps
    cmap = colormaps[i % len(colormaps)]
    
    # Plot using geopandas with thread safety
    with plot_lock:
        sampled_gdf.plot(column=pc_col, cmap=cmap, ax=axes[i], 
                         markersize=0.5, rasterized=True)
        axes[i].set_title(title)
        axes[i].set_xlabel('Longitude')
        axes[i].set_ylabel('Latitude')
        axes[i].grid(True, alpha=0.3)

# Generate plots in parallel using ThreadPoolExecutor
with ThreadPoolExecutor(max_workers=9) as executor:
    futures = [executor.submit(create_plot, i, pc_col) for i, pc_col in enumerate(pc_columns)]
    # Wait for all plots to complete
    for future in futures:
        future.result()

plt.tight_layout()
plt.savefig("pca_plots.png", dpi=100)
```
